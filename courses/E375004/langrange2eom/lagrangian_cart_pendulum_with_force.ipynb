# 📓 Lagrangian Mechanics Tutorial
## 🔹 Example 3: Pendulum on a Cart with Gravity and External Force

```
from sympy import symbols, Function, diff, simplify, solve, Eq, dsolve, sin, cos
from sympy.physics.mechanics import dynamicsymbols
from sympy import init_printing
from IPython.display import display

init_printing()
```

## 🔹 1. Define Coordinates and Time
Let the cart move along a horizontal track with position $x(t)$. The pendulum is attached to the cart and swings with angle $\theta(t)$ from the vertical (downward).

```
t = symbols('t')
x = dynamicsymbols('x')             # Cart position
theta = dynamicsymbols('theta')     # Pendulum angle from vertical

m, M, l, g, F = symbols('m M l g F')  # pendulum mass, cart mass, pendulum length, gravity, external force
```

## 🔹 2. Kinetic Energy $T$
The cart and pendulum both contribute to the kinetic energy. The pendulum's tip velocity has both horizontal and vertical components.

$$
T = \frac{1}{2} M \dot{x}^2 + \frac{1}{2} m \left[ (\dot{x} + l \dot{\theta} \cos\theta)^2 + (l \dot{\theta} \sin\theta)^2 \right]
$$

```
x_dot = diff(x, t)
theta_dot = diff(theta, t)

# Velocity components of pendulum mass center
vx = x_dot + l * theta_dot * cos(theta)
vy = l * theta_dot * sin(theta)

T = (1/2) * M * x_dot**2 + (1/2) * m * (vx**2 + vy**2)
T = simplify(T)
display(T)
```

## 🔹 3. Potential Energy $V$
The cart is on a flat surface. Only the pendulum contributes potential energy:

$$
V = -m g l \cos\theta
$$

```
V = -m * g * l * cos(theta)
display(V)
```

## 🔹 4. Lagrangian $L = T - V$

```
L = T - V
display(L)
```

## 🔹 5. Apply Euler-Lagrange Equations
We apply the equations separately for $x$ and $\theta$:

For $x$ (with external force $F$):

$$
\frac{d}{dt} \left( \frac{\partial L}{\partial \dot{x}} \right) - \frac{\partial L}{\partial x} = F
$$

For $\theta$:

$$
\frac{d}{dt} \left( \frac{\partial L}{\partial \dot{\theta}} \right) - \frac{\partial L}{\partial \theta} = 0
$$

```
# For x
dL_dx = diff(L, x)
dL_dx_dot = diff(L, x_dot)
d_dt_dL_dx_dot = diff(dL_dx_dot, t)
EL_x = simplify(d_dt_dL_dx_dot - dL_dx - F)
display(EL_x)
```

```
# For theta
dL_dtheta = diff(L, theta)
dL_dtheta_dot = diff(L, theta_dot)
d_dt_dL_dtheta_dot = diff(dL_dtheta_dot, t)
EL_theta = simplify(d_dt_dL_dtheta_dot - dL_dtheta)
display(EL_theta)
```

## ✅ Final Result
These are the coupled nonlinear equations of motion for the cart-pendulum system with external force $F$ and gravity.

## 🌟 Conclusion
We have successfully derived the full nonlinear dynamics of a pendulum on a cart using Lagrangian mechanics and symbolic computation.
