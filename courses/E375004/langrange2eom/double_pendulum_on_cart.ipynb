
# Derivation of Equations of Motion for a Double Pendulum on a Cart using Lagrange's Equation

## Introduction

This notebook demonstrates the step-by-step derivation of the equations of motion for a double pendulum attached to a cart using Lagrange's equation.

## Setup

First, we import necessary libraries and define our symbols.
    

```

import numpy as np
import sympy as sp
from sympy import symbols, diff, sin, cos

sp.init_printing()

# Define symbols
x, theta1, theta2, l1, l2, m1, m2, M, g = symbols('x theta1 theta2 l1 l2 m1 m2 M g')
xdot, theta1dot, theta2dot = symbols('xdot theta1dot theta2dot')
xddot, theta1ddot, theta2ddot = symbols('xddot theta1ddot theta2ddot')
    
```


## Kinetic and Potential Energy

We calculate the kinetic and potential energy of the system.
    

```

# Positions of the masses
x1 = x + l1*sin(theta1)
y1 = -l1*cos(theta1)
x2 = x1 + l2*sin(theta2)
y2 = y1 - l2*cos(theta2)

# Velocities of the masses
v1_squared = (diff(x1, x)*xdot + diff(x1, theta1)*theta1dot)**2 + (diff(y1, x)*xdot + diff(y1, theta1)*theta1dot)**2
v2_squared = (diff(x2, x)*xdot + diff(x2, theta1)*theta1dot + diff(x2, theta2)*theta2dot)**2 + (diff(y2, x)*xdot + diff(y2, theta1)*theta1dot + diff(y2, theta2)*theta2dot)**2

# Kinetic Energy
T_cart = 1/2 * M * xdot**2
T_m1 = 1/2 * m1 * v1_squared
T_m2 = 1/2 * m2 * v2_squared
T_total = T_cart + T_m1 + T_m2

# Potential Energy
V_m1 = m1 * g * y1
V_m2 = m2 * g * y2
V_total = V_m1 + V_m2
    
```


## Lagrangian and Equations of Motion

We form the Lagrangian and derive the equations of motion.
    

```

# Lagrangian
L = T_total - V_total

# Equations of Motion
eq_x = diff(diff(L, xdot), 't') - diff(L, x)
eq_theta1 = diff(diff(L, theta1dot), 't') - diff(L, theta1)
eq_theta2 = diff(diff(L, theta2dot), 't') - diff(L, theta2)

# Display equations (for demonstration, might not solve analytically here)
eq_x, eq_theta1, eq_theta2
    
```
